# **1. Software Development Methodologies**

This is a VERY important interview topic because interviewers want to know:

> “Do you understand how real software teams organize development?”

Especially for AI engineers working in:

* Startups
* Product companies
* Enterprise systems
* Research-to-production environments

## **Interview-Level Definition**

Software Development Methodologies are structured approaches used to plan, manage, and execute software development.

### **Simple Definition**

“A methodology is the process or strategy a team follows to build software.”

It defines:

* How work is organized
* How teams collaborate
* How requirements are handled
* How testing/deployment happen

# **2. Why Methodologies Are Needed**

Without methodologies:

* Teams become chaotic
* Deadlines fail
* Bugs increase
* Requirements get lost
* Collaboration becomes difficult

Methodologies provide:

* Structure
* Predictability
* Communication
* Quality control

# **4. Waterfall Model**

What is it?

> A linear sequential methodology.

Each phase happens once:

* Requirements
* Design
* Development
* Testing
* Deployment
* Maintenance

> How it works: You complete one phase fully before moving to next.

**Advantages:**

* Simple
* Structured
* Easy documentation
* Clear milestones

**Disadvantages:**
* Hard to change requirements
* Feedback comes late
* High risk if requirements wrong

**Best for:**
* Fixed-scope projects
* Government systems
* Highly regulated industries

**Interview insight**

“Waterfall works best when requirements are stable and predictable.”

# **7. Kanban**

> A visual workflow methodology focused on continuous delivery.

**Workflow Board**

| To Do | In Progress | Done |
| ----- | ----------- | ---- |

**Key Idea**

Limit work-in-progress to improve flow.

> Advantages
* Simple
* Flexible
* Great for maintenance/support teams

> Best for
* DevOps
* Monitoring teams
* AI inference operations


# **8. Spiral Model**

> Risk-driven iterative methodology.

Combines:

* Iteration
* Risk analysis
* Prototyping

> Steps in Spiral
* Planning
* Risk analysis
* Engineering
* Evaluation

Repeat in cycles.

 Software aik hi baar me nahi banta.
Rounds/cycles me banta hai.

👉 build → test → improve → repeat

> Risk-driven

Har cycle me pehle biggest risks identify karte hain.

Example:

* AI model accurate hoga ya nahi?
* system scale karega?
* security issue hai?

Pehle risky cheezen test karte hain.

> 🌀 Why “spiral”?

Because process circles/loops me repeat hota rehta hai:

* planning
* risk analysis
* development
* testing
* then next loop

> 🧠 Easiest memory trick:

Spiral Model =

👉 “Build little by little while continuously checking risks.”


> Advantages
* Handles high-risk projects well
* Good for large systems

> Disadvantages
* Complex
* Expensive
* Requires expertise

> Best for
* Aerospace
* Banking
* Mission-critical systems

# **10. Lean Development**

Inspired by Toyota manufacturing.

> Focus:

* Eliminate waste
* Deliver value quickly
* Continuous improvement

> Waste examples in software:
* Unnecessary features
* Waiting time
* Rework
* Excess documentation

🔹 Lean Development kya hai?

Lean development ka focus hota hai:

👉 “Kam waste karo, jaldi value deliver karo.”

Yeh Toyota manufacturing se inspired hai.

> 🔥 Main idea:

Sirf woh kaam karo jo actually user/business ko value deta hai.
Extra unnecessary cheezen avoid karo.

🗑️ Waste examples in software

**❌ Unnecessary features**
Aisi features banana jo users use hi nahi karte.


**⏳ Waiting time**
Developers ka approvals, testing, ya dependencies ki wajah se idle rehna.

**🔁 Rework**
Galat requirements ki wajah se same kaam dobara karna.

**📄 Excess documentation**
Itni documentation likhna jo practically koi use hi nahi karta.

**🔥 Interview line:**

“Lean development focuses on eliminating waste, improving efficiency, and delivering customer value quickly through continuous improvement.”

# **11. RAD (Rapid Application Development)**

> Focus:

* Rapid prototyping
* Fast user feedback
* Quick iterations

> Best for
* UI-heavy apps
* Fast MVPs
* Startup prototyping

🔹 RAD (Rapid Application Development) kya hai?

RAD ek software development model hai jiska focus hota hai:

👉 “Software jaldi banao, user ko dikhao, feedback lo, aur quickly improve karo.”

> 🔥 Main focus:

**⚡ Rapid prototyping**

Pehle quickly basic working version/prototype bana dete hain.

**👤 Fast user feedback**

User se jaldi feedback lete hain instead of months wait karna.

**🔄 Quick iterations**

Feedback ke basis pe fast changes/improvements karte hain.

> ✅ Best for:

**🎨 UI-heavy apps**

Jahan interface aur user experience important ho.

**🚀 Fast MVPs**

Startup products ya demo versions quickly launch karne ke liye.

**💡 Startup prototyping**

Idea jaldi validate karna without full complex development.

### **Difference in RAD and Agile:**

**🧠 Easy memory trick:**

Agile = “flexible iterative development”

RAD = “rapid prototype-based development”

**🔥 Real-world feel:**

Agile → proper product development approach

RAD → “jaldi MVP/demo bana do”

# **------------ Appendices --------------**

##**🌀 When would you use Spiral model?**

Spiral model is used in large or high-risk projects where requirements are unclear or risks are significant. It helps by developing the system in cycles while continuously analyzing risks.

##**📋 Difference between Kanban and Scrum?**

Scrum uses fixed sprints and planned work cycles, while Kanban has continuous workflow without fixed sprint durations. Scrum is more structured; Kanban is more flexible.

##**⚙️ Why is DevOps important?**
DevOps improves collaboration between development and operations teams and enables faster, reliable software delivery through automation, CI/CD, and monitoring.

##**🤖 Why are AI projects naturally iterative?**
AI projects depend heavily on data and model performance, so models need continuous training, testing, tuning, and improvement based on new data and feedback.

##**🔄 Explain MLOps in context of DevOps.**
MLOps is DevOps for machine learning systems. It automates the ML lifecycle including data pipelines, model training, deployment, monitoring, and retraining in production.

### **🔥 Scrum vs Kanban**

## **🔹 Scrum**

Kaam fixed time periods me hota hai called Sprints.

Example:

“Next 2 weeks me yeh 5 tasks complete karne hain.”

Sprint start hone ke baad usually tasks change nahi karte.

👉 Structured system

## **🔹 Kanban**

No sprints.
Kaam continuously flow karta rehta hai.

Board hota hai:

* To Do
* Doing
* Done

Jab new task aaye → directly add kar do.

👉 Flexible continuous workflow

Query: hmary to office me agile method chalta eh. kanban board jira use krty hain hm or sprints me kaam bhejty hainrealse me

Answer: Real-world teams often use a hybrid approach instead of “pure Scrum” or “pure Kanban”.

Your team is basically doing:

✅ Scrum-style sprints

✅ Kanban-style Jira board

This is normal.

Why?

Because:

Scrum gives structure through sprints/planning
Kanban board makes task tracking easy visually

So companies combine both.

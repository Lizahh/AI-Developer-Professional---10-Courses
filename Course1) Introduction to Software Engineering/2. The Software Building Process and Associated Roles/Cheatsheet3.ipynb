# **Software Versions**

This is a VERY important real-world engineering topic because every serious software system evolves over time.

Interviewers want to know whether you understand:

* How software changes are managed
* How teams release updates safely
* How compatibility is maintained
* How production systems evolve

Especially important for:

* AI systems
* APIs
* SaaS products
* Mobile/web applications
* ML models

# **1. What Are Software Versions?**

> Interview-Level Definition

A software version is a unique identifier assigned to a specific state or release of software.

> Simple Definition

“Software versions help track changes, updates, improvements, and releases over time.”

## **Why Versions Matter**

Without versions:

* Teams get confused
* Deployments become risky
* Bugs are hard to track
* Compatibility breaks
* Rollbacks become difficult

> Real-World Example

* ChatGPT v1
* ChatGPT v2
* GPT-4
* GPT-5

> Each version:

* Adds improvements
* Fixes issues
* Changes functionality

# **2. Why Software Changes Over Time**

Software is never truly “finished.”

> Reasons:

* Bugs discovered
* Security vulnerabilities
* New features needed
* Performance optimization
* Scaling requirements
* User feedback
* Technology evolution

> Interview Gold Line

“Software engineering is continuous evolution, not one-time development.”

# **3. Version Numbering (VERY IMPORTANT)**

> Most companies use:

Semantic Versioning (SemVer)

> Format:

MAJOR.MINOR.PATCH

> Example:

2.5.1

# **4. Meaning of Each Part**

## **A. MAJOR Version**

> Changes when:

* Breaking changes occur
* APIs become incompatible
* Major redesign happens

> Example:

1.0.0 → 2.0.0

> Example

* Python 2 → Python 3
* Major compatibility changes

## **B. MINOR Version**

Changes when:

* New features added
* Backward compatibility maintained

> Example:

2.4.0 → 2.5.0

## **C. PATCH Version**

Changes when:

* Bugs fixed
* Small improvements
* Security patches

Example:

2.5.0 → 2.5.1

# **Interview Insight**

| Change Type     | Version Updated |
| --------------- | --------------- |
| Bug fix         | PATCH           |
| New feature     | MINOR           |
| Breaking change | MAJOR           |


# **5. Types of Software Releases**

## **A. Alpha Release**

Internal testing phase.

> Characteristics:

* Incomplete
* Unstable
* Developers/testers only

## **B. Beta Release**

Released to limited users.

> Purpose:

* Gather feedback
* Identify bugs



## **C. Stable Release**

Production-ready version.

## **D. Hotfix**

Urgent patch for critical issues.

> Example:

Security vulnerability fix



## **E. LTS (Long-Term Support)**

Stable versions supported for years.

> Examples:

* Ubuntu LTS
* Node.js LTS

# **6. Version Control vs Software Versions (IMPORTANT)**

Students confuse these often.

## **Software Version**

Represents product release.

Example:

v2.1.0
## **Version Control**

Tracks code changes internally.

Tools:

* Git
* GitHub
* GitLab

# **7. Backward Compatibility**

VERY IMPORTANT interview concept.

"New versions should not break old systems."

> Example:

Old API clients should still work after upgrade.

> Why Important?

Breaking compatibility:

* Frustrates users
* Breaks integrations
* Causes outages

> Example

* Bad:
Removing API fields suddenly.

* Better:
Deprecate old fields gradually.

# **8. API Versioning (VERY IMPORTANT)**

Modern systems version APIs carefully. API versioning ka matlab hai same API ko different versions me run karna taake purane aur naye systems dono smoothly kaam karein.

## **Common API Versioning**
* /api/v1/users
* /api/v2/users

👉 v1 purani functionality
👉 v2 updated/new features

> Why?

Allows:

* Old clients to continue working
* New features without breaking integrations
* Old apps break nahi hoti
* New features add ho jate hain safely
* Backward compatibility maintain rehti hai

> AI Example

ML inference API:

* v1 = sirf text classification karta hai
* v2 = text + images (multimodal AI)

# **Software Testing**

“Testing checks whether software behaves as expected.”

> Interview gold line:

“Testing increases confidence in software quality; it does not guarantee absence of bugs.”

# **1. Types of Software Testing (VERY IMPORTANT)**


## **A. Unit Testing**

Tests individual functions/components in isolation.

> Purpose
* Catch small logic errors early
* Ensure individual modules work correctly

## **B. Integration Testing**

Tests whether multiple components work together properly.

> Example

Testing:

* Backend API + Database
* AI model + inference API

> Purpose

* Detect communication failures between modules.

## **C. System Testing**

Tests the complete system as a whole.

## **D. Acceptance Testing (UAT)**

Validates whether software meets business/user requirements.

> Performed by:

* Clients
* Product teams
* End users

## **E. Regression Testing**

Ensures new changes do NOT break existing functionality.

* Modern software changes constantly.
* Regression testing prevents old features from failing.

> Example

* After adding multilingual support:

* Login system should still work

## **F. Performance Testing**

Tests:

* Speed
* Stability
* Scalability

> Types

* Load Testing

* Expected traffic.

* Stress Testing

* Extreme traffic.

* Spike Testing

* Sudden traffic increases.

> AI Example

Can inference API handle:

* 100K requests/minute?

## **G. Security Testing**

Checks:

* Authentication
* Authorization
* Vulnerabilities
* Data protection

## **H. Usability Testing**

Tests:

* User experience
* Interface clarity
* Accessibility


# **2. Manual vs Automated Testing**

> 1. Manual Testing

Humans test system manually.

> Advantages
* Good for UI/UX
* Exploratory testing

> Disadvantages
* Slow
* Repetitive
* Human error

> 2. Automated Testing

Tests run automatically using tools/scripts.

> Advantages
* Fast
* Repeatable
* Scalable
* CI/CD friendly

> Common Tools
* PyTest
* JUnit
* Selenium
* Cypress


# **3.Agile Testing Principle**

“Testing is not a separate final phase—it happens continuously.”

# **4. CI/CD and Testing (VERY IMPORTANT)**

Modern systems use:

* Continuous Integration
* Continuous Deployment

> Workflow

* Developer pushes code →
* Automated tests run →
* Deployment happens if tests pass


# **5. Software Testing in AI Systems (VERY IMPORTANT)**

AI testing is DIFFERENT from traditional testing.

Why?

> Traditional software:

* Deterministic outputs

> AI systems:

* Probabilistic outputs






# **6. AI-Specific Testing**

## **A. Model Accuracy Testing**

Metrics:

* Accuracy
* Precision
* Recall
* F1-score

## **B. Bias Testing**

Checks fairness across:

* Demographics
* Data distributions

## **C. Drift Detection**

Model performance may degrade over time.

## **D. Latency Testing**

Inference response time matters.

> Example:

Real-time fraud detection systems

## **E. Data Validation**

Check:

* Missing values
* Data schema
* Distribution shifts

# **7. Common Testing Strategies**

> 1. Test-Driven Development (TDD)


Write tests BEFORE writing code.

> Cycle:

* Write test
* Fail test
* Write code
* Pass test

> Benefits
* Better design
* Cleaner code
* Safer refactoring

> 2. Behavior-Driven Development (BDD)

Focus:

User behavior scenarios

❌ Mistake 1:

Thinking testing only means finding bugs

✔ Reality:

Testing validates quality, reliability, scalability, and behavior.

❌ Mistake 2:

Assuming AI testing is same as traditional testing

✔ Reality:

AI outputs are probabilistic and data-dependent.

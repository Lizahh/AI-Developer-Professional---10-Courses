# **Building Quality Software**

This topic is EXTREMELY important in interviews because companies care less about:

> “Can you write code?”

and more about:

> “Can you build reliable, maintainable, production-quality software?”

Especially for AI engineers.

# **1. What Is Quality Software?**

Quality software is software that:

* Works correctly
* Meets user requirements
* Is reliable
* Is maintainable
* Is secure
* Performs efficiently
* Can evolve over time
* Simple Human Definition

> “Good software is not just software that works today — it continues working reliably as systems grow and change.”

## **Key Characteristics of Quality Software**

These are HIGHLY important for interviews.

**1. Correctness**

Software should produce correct outputs.

> Example:

* Banking software must calculate balances correctly
* AI classifier should predict accurately

**2. Reliability**

Software should work consistently under different conditions.

> Example:

* System should not crash under traffic spikes
* APIs should handle failures gracefully

**3. Maintainability**

Code should be easy to:

* Understand
* Modify
* Debug
* Extend

Interview gold insight:

* “Most software engineering effort goes into maintenance.”

**4. Scalability**

Software should handle growth:

* More users
* More requests
* More data

AI example:

> LLM inference systems scaling to millions of requests

**5. Security**

Protect:

* User data
* APIs
* Infrastructure

> Examples:

* Authentication
* Encryption
* Authorization

**6. Performance**

Efficient use of:

* CPU
* Memory
* GPU
* Network

AI relevance:

* Inference latency optimization
* Vector DB retrieval speed

**7. Usability**

Software should be easy for users to understand and use.

> Real Industry Truth

Many beginners think:

“If code runs, software is complete.”

In reality:

Working code can still be terrible software.

Example:

* No tests
* Poor structure
* Impossible to scale
* Hard to debug
* Insecure

> Example: Poor vs Quality Software

Poor Software

            def x(a,b,c):
                return a*b/c

Problems:

* Bad naming
* No validation
* No documentation
* Division-by-zero risk


> Better Quality Software

      def calculate_average_price(total_price, quantity, days):
          """
          Calculate average price per day.
          """

          if days == 0:
              raise ValueError("Days cannot be zero")

          return (total_price * quantity) / days

Better because:

* Readable
* Safe
* Maintainable
* Self-documenting

> Quality in AI Systems (VERY IMPORTANT)

AI engineers often focus only on:

* Accuracy
* Training

But production AI quality also includes:

* AI Concern	Software Engineering Concern
* Model accuracy	API reliability
* Training	Deployment
* Hyperparameters	Scalability
* Datasets	Monitoring
* Experiments	Maintainability

> Example: AI Chatbot

Bad AI software:

* Accurate model
* But API crashes frequently

Good AI software:

* Stable API
* Logging
* Monitoring
* Fast responses
* Retry mechanisms
* Graceful error handling

> Example:

* “Fast system” is vague.

* “Response time must be under 200ms.”

2. Good System Design

> Examples:

* Modular systems
* Separation of concerns
* Scalable infrastructure

3. Clean Coding Practices

Includes:

* Readable naming
* Consistent structure
* Reusable modules
* Error handling

4. Version Control (Git)

Helps:

* Collaboration
* Tracking changes
* Rollbacks

> Industry truth:

No serious engineering happens without version control.

5. Code Reviews

Senior engineers review code before merging.

Purpose:

* Catch bugs
* Improve readability
* Enforce standards
* Share knowledge

> Interview insight:

“Code reviews improve both code quality and team learning.”

# **6. Testing**

One of the BIGGEST quality drivers.

* Types of Testing

**1. Unit Testing**

* Tests small functions/modules.

**2. Integration Testing**

Tests components together.

👉 Focus: components ka interaction

**3. System Testing**

Tests complete application.

👉 Focus: full app as a whole

**4. Performance Testing**

Checks scalability and load handling.

**5. Regression Testing**

“Naya change aane ke baad purana sab abhi bhi kaam kar raha hai ya nahi?”

👉 Focus: old features not broken

> Example:

* tumne new feature add kiya
* ab check karo login/payment/screen pehle jaisa hi chal raha hai?

👉 Focus: old features not broken

🧠 Super easy trick yaad rakhne ka:

* Integration = parts connect ho rahe hain? (ALL ABOUT NEW SYSTEM)
* System = poora system sahi chal raha hai?(ALL ABOUT NEW SYSTEM)
* Regression = purana kuch toot to nahi gaya? (OLD+NEW SYSTEM)


**6. AI-Specific Testing**

* Accuracy evaluation
* Drift detection
* Bias analysis
* Latency testing

# **Requirements (Software Engineering)**
This is one of the MOST critical topics in interviews because many system failures happen due to bad or misunderstood requirements, not bad coding.
For AI engineers especially, this is where most real-world mistakes begin.

## **1. What are Requirements?**

> Interview-level definition:

“What the software should do and what constraints it must satisfy.”

> Simple definition:

“Requirements are the agreement between what users want and what engineers will build.”

## **2. Why Requirements are Important**
Interviewers want to see if you understand:

1. Prevents wrong building

If requirements are wrong → you build the wrong system correctly.

2. Reduces rework

Clear requirements save time and cost.

3. Aligns stakeholders

Ensures everyone agrees on what is being built.

4. Foundation of SDLC

Everything starts from requirements:

•	Design

•	Development

•	Testing

•	Deployment

All depend on it.

## **3. Types of Requirements**
There are TWO main types:

### **A. Functional Requirements (WHAT system does)**
These define system behavior.

Functional requirements are the features of the system.

> Examples:

•	User can login

•	User can upload file

•	System sends email notification

•	AI model predicts sentiment

> AI example:

For a recommendation system:

•	Suggest top 10 products

•	Filter based on user history

•	Update recommendations daily

### **B. Non-Functional Requirements (HOW system performs)**

These define system quality and constraints.

“Non-functional requirements often matter more in production than functional ones.”

> Examples:

•	Performance

•	Scalability

•	Security

•	Reliability

•	Availability

> AI example:

•	Response time < 200ms

•	System handles 1M requests/day

•	Model accuracy > 90%

•	99.9% uptime

## **5. Requirement Gathering Process**

This is how real teams define requirements:

**Step 1: Understand problem**

Talk to stakeholders:

•	Product managers

•	Users

•	Business teams

**Step 2: Ask clarifying questions**

•	What is priority?

•	What is expected load?

•	What are constraints?

**Step 3: Document requirements**

Write SRS (Software Requirement Specification)

**Step 4: Validate requirements**

Make sure:

•	Nothing is missing

•	No contradictions

•	Realistic expectations

**Step 5: Final approval**

Stakeholders agree before development starts

## **6. Common Problems in Requirements**
**1. Ambiguity**

> Bad:

•	“System should be fast”
> Good:

•	“Response time should be < 200ms”

**2. Incompleteness**

> Missing:

•	Edge cases

•	Failure handling

•	Load expectations

**3. Changing requirements**

Very common in real projects:

•	New features added mid-development

**4. Conflicting requirements**

Example:

•	“High security” + “No login required”

Impossible combination.

## **7. Requirements in Agile vs Waterfall**

> Waterfall:

•	Requirements defined fully at start

•	Hard to change later

> Agile:

•	Requirements evolve continuously

•	Updated every sprint

> Interview insight:

“Modern AI systems prefer Agile because requirements evolve with data and experiments.”

## **8. Requirements in AI Systems (VERY IMPORTANT)**

AI systems have UNIQUE requirements:

**1. Data requirements**

•	What data is needed?

•	How much data?

•	Is it labeled?

**2. Model performance requirements**

•	Accuracy

•	Precision/recall

•	F1 score

**3. Latency requirements**

•	Real-time vs batch inference

**4. Infrastructure requirements**

•	GPU/CPU needs

•	Memory requirements

**5. Monitoring requirements**

•	Model drift detection

•	Performance tracking

### **Example: Fraud Detection System**

**Functional:**

•	Detect fraudulent transactions

•	Flag suspicious activity

**Non-functional:**

•	Must respond in < 100ms

•	99.99% accuracy goal (hard in reality → tradeoff discussion)

•	Handle millions of transactions/day

## **9. Good Requirement Writing (Interview Tip)**
Good requirements are:

**SMART:**

•	Specific

•	Measurable

•	Achievable

•	Relevant

•	Time-bound

**Example:**

> Bad:

•	“System should be secure”

> Good:

•	“All user data must be encrypted using AES-256 and access controlled via role-based authentication.”

# **--------- Appendices ---------**

## **🔍 How do code reviews improve quality?**

Code reviews help catch bugs, bad design, and security issues before production. They also improve code readability and ensure best practices are followed across the team.

## **⚙️ Difference between functional and non-functional quality?**

Functional quality means what the system does (features working correctly).
Non-functional quality means how well it performs (speed, scalability, security, reliability).

## **📊 Why is monitoring important?**

Monitoring helps detect issues in real-time like errors, latency, or system failures. It ensures we can fix problems quickly before users are heavily affected.

# **1. Introduction to the SDLC**

## **What is SDLC?**

The Software Development Life Cycle (SDLC) is a structured process used to design, develop, test, deploy, and maintain software systems.

**Simple definition:**

“SDLC is the step-by-step journey of turning an idea into a working software product and keeping it alive in production.”

## **Why SDLC exists**

> Without SDLC:

* Teams build randomly
* Requirements get lost
* Bugs increase
* Systems break in production
* Scaling becomes impossible

> With SDLC:

* Clear process
* Better quality
* Predictable delivery
* Easier maintenance
* Team coordination
* Interview-level insight

## **SDLC in AI systems**

Example: Building an AI chatbot

> SDLC ensures:

* Requirements: What should chatbot do?
* Design: API + model architecture
* Development: training + backend code
* Testing: accuracy + load testing
* Deployment: serving model via API
* Maintenance: retraining, monitoring drift

# **2. Phases of SDLC**

There are 6–7 standard phases depending on company, but interviews expect this structure:

### **1. Requirement Analysis**

Understanding what the system should do.

* Key activities:
* Talking to stakeholders
* Defining features
* Identifying constraints
* Writing requirements document

> Output:

Software Requirement Specification (SRS)

> AI example:

For a recommendation system:

* Should recommend products?
* Real-time or batch?
* Personalized or global?

### **2. System Design**

Planning how the system will work internally.

Includes:
* Architecture design
* Database design
* API design
* Tech stack selection

> Output:

* High-Level Design (HLD)
* Low-Level Design (LLD)

> AI example:

* Model serving architecture
* Vector database design
* Microservices for inference

### **3. Implementation (Coding Phase)**

Actual development of the system.

> Activities:
* Writing code
* Building APIs
* Training ML models
* Integrating services

> Important:
* Follows design docs
* Uses version control (Git)

“Good engineering is writing code that follows design, not just working code.”

### **4. Testing**

Ensuring software works correctly.

> Types:
* Unit testing (individual functions)
* Integration testing (services together)
* System testing (whole system)
* Performance testing (load, stress)

> AI-specific testing:
* Model accuracy testing
* Bias testing
* Latency testing


### **5. Deployment**

Moving software to production.

> Includes:
* CI/CD pipelines
* Docker containers
* Cloud deployment (AWS, GCP, Azure)

> AI example:
* Deploying model as REST API
* GPU-based inference service

6. Maintenance

Ongoing updates and fixes after release.

> Activities:
* Bug fixing
* Performance improvements
* Feature updates
* Model retraining (AI systems)

“Most of a system’s life is spent in maintenance, not development.”

### **Optional Phase (Advanced in some companies)**
### **7. Monitoring & Feedback**

Modern systems include:

* Logging
* Monitoring dashboards
* Error tracking
* User feedback loops

> AI systems:

* Detect model drift
* Track prediction quality
* Trigger retraining pipelines

# **3. Activity: Phases in the SDLC**

Let’s apply SDLC to a real AI engineering scenario.

### **Scenario: Build a Spam Email Detection System**

1. Requirement Analysis

* Detect spam emails
* High accuracy required
* Real-time classification
* Handle millions of emails/day

2. System Design
* Input: email text
* Model: NLP classifier
* API service for prediction
* Database for logs
* Queue for email processing

3. Implementation
* Train NLP model (BERT / TF-IDF + ML)
* Build backend API (FastAPI/Django)
* Integrate model into service

4. Testing
* Test classification accuracy
* Test false positives/negatives
* Load test API

5. Deployment
* Deploy API on cloud
* Containerize using Docker
* Use load balancer for traffic

6. Maintenance
* Monitor spam detection accuracy
* Retrain model with new data
* Improve performance over time

7. Monitoring (optional but real-world)
* Track incoming email patterns
* Detect new spam techniques
* Alert system for anomalies

## **🤔 How do you handle unclear requirements?**

I clarify requirements by asking targeted questions, defining assumptions, and breaking the problem into smaller parts. If details are still unclear, I start with a minimal prototype and iterate with feedback.

## **🧪 Why is testing important?**
Testing ensures the system works correctly, catches bugs early, and prevents failures in production. It also builds confidence that new changes don’t break existing functionality.

## **🔄 How does SDLC apply to ML systems?**
SDLC in ML includes not just coding, but also data collection, model training, evaluation, deployment, and continuous monitoring. It’s a cycle because models need updates as data changes.

## **🤖 What is different in AI SDLC vs normal software SDLC?**
Normal SDLC focuses on code behavior, while AI SDLC depends heavily on data and model performance. In AI, output is probabilistic, so continuous monitoring and retraining are needed.

## **🧠 Where does model training fit in SDLC?**
Model training is part of the development phase (between data preparation and evaluation). After training, the model is tested, validated, and then deployed into production.

# **---------------- Agile Methodology ---------------**

## **1. What is Agile Methodology?**

Agile is a software development approach where software is built in small, iterative cycles instead of one long, rigid plan.

Simple definition:

“Agile is a way of building software step-by-step with continuous feedback and adaptation.”

> Instead of:

planning everything upfront ❌ (Waterfall style)

> Agile does:

build small → test → get feedback → improve → repeat ✔

## **2. Why Agile was needed (VERY IMPORTANT for interviews)**

Before Agile, companies used the Waterfall model:

* Requirements → Design → Build → Test → Deploy (one long sequence)

> Problem with Waterfall:

* Requirements change often ❌
* Users don’t know what they want early ❌
* Bugs are discovered too late ❌
* Delivery takes too long ❌

> Real-world issue:

By the time software is delivered, the market already changed.

So Agile was created to solve:

1. Changing requirements: Software evolves constantly.

2. Faster delivery: Users want features quickly.

3. Feedback loops: Build → test → improve continuously.

4. Risk reduction: Small releases reduce failure impact.

## **3. Core Idea of Agile**

“Deliver working software frequently, in small increments, and continuously improve it.”

## **4. How Agile Works (Step-by-Step)**

Agile is not one process—it is a framework of iterative cycles.

**Step 1: Break work into small pieces**

> Instead of building everything at once:

* Split into features (user stories)

> Example:

* AI chatbot project:

    * Login system
    * Chat API
    * Model inference
    * History storage

**Step 2: Work in Iterations (Sprints)**

A sprint = short development cycle (usually 1–2 weeks)

Each sprint includes:

* Planning
* Development
* Testing
* Delivery

**Step 3: Build → Test → Feedback loop**

After each sprint:

* Show working software
* Get feedback from users/stakeholders
* Improve next sprint

**Step 4: Repeat**

This cycle continues until product is complete (and even after release).


## **5. Agile Principles (from Agile Manifesto)**

These are VERY important for interviews.

**1. Individuals and interactions over processes and tools**

People matter more than strict tools.

**2. Working software over documentation**

Focus on product, not paperwork.

**3. Customer collaboration over contract negotiation**

Continuous user feedback is important.

**4. Responding to change over following a plan**

Adapt instead of rigid execution.

# **6. Agile vs Waterfall (Common Interview Question)**

| Feature     | Agile                   | Waterfall            |
| ----------- | ----------------------- | -------------------- |
| Approach    | Iterative               | Sequential           |
| Flexibility | High                    | Low                  |
| Delivery    | Frequent                | One-time             |
| Feedback    | Continuous              | Late                 |
| Risk        | Low                     | High                 |
| Best for    | Modern apps, AI systems | Fixed-scope projects |



# **7. Agile Frameworks (Important for interviews)**

Agile is umbrella term. Common frameworks:

### **1. Scrum (MOST IMPORTANT)**

Scrum is the most widely used Agile framework.

> Key roles:

* Product Owner → defines requirements
* Scrum Master → removes blockers
* Development Team → builds product

> Scrum cycle:

* Sprint Planning
* Daily Standups
* Sprint Development
* Sprint Review
* Sprint Retrospective

### **2. Kanban**

Focus: continuous flow of work

> Uses board:

* To Do
* In Progress
* Done

> Used in:

* Maintenance teams
* DevOps
* AI model monitoring pipelines

### **3. XP (Extreme Programming)**

Focus:

* Pair programming
* Continuous testing
* Frequent releases

### **Example (Netflix / Amazon style thinking):**

Release feature → measure usage → improve → scale

# **9. Agile in AI Engineering (VERY IMPORTANT)**

AI systems are PERFECT for Agile because:

1. Models are iterative

Train → evaluate → improve → retrain

2. Data changes over time

Data drift requires updates

3. Experimentation is constant

Different models, hyperparameters, architectures

> **Example: AI Recommendation System**

Sprint 1:

Basic model (collaborative filtering)

Sprint 2:

Add content-based filtering

Sprint 3:

Add deep learning model

Sprint 4:

Add real-time inference

Each sprint improves system incrementally.

# **10. Disadvantages of Agile (interviewers LOVE this)**

* Hard to predict final cost/time
* Requires strong team communication
* Less documentation
* Scope can keep changing
* Not ideal for very rigid regulated systems

# **11. Common Mistakes Candidates Make**

❌ Mistake 1:

Saying Agile = Scrum (wrong)

✔ Agile = philosophy
✔ Scrum/Kanban = frameworks

❌ Mistake 2:

Thinking Agile has no planning

✔ Agile still has planning—just iterative planning

❌ Mistake 3:

Ignoring testing in Agile

✔ Testing happens in every sprint

## **⚡ Load testing**

System ko normal expected traffic pe test karna.
Check karte hain system kitne users smoothly handle kar sakta hai.

## **🔥 Stress testing**

System ko extreme / beyond limit load dena.
Dekhte hain system crash hota hai ya gracefully fail karta hai.

## **🧪 Regression testing**

New changes ke baad check karna ke purana functionality break to nahi hui.
👉 “old features still working?” verify karna

## **🎯 Model accuracy testing**

ML model kitna correct predictions de raha hai, usko measure karna.
Usually accuracy, precision, recall se check hota hai.

## **⚖️ Bias testing**

Check karna ke model kisi group (gender, race, etc.) ko unfair treat to nahi kar raha.
👉 Fairness of model evaluate karna

## **⏱️ Latency testing**

Ek request ko process hone me kitna time lagta hai, wo measure karna.
👉 “response kitni fast aati hai” check karna

* Feature = functionality/product capability
> Example: “User login system”

* User story = small requirement written from user perspective
>  Example: “As a user, I want to log in using email and password so I can access my account.”

In Agile:

👉 Big features are usually broken into smaller user stories.

So:
Feature → multiple user stories → tasks

# **1. What Is Software Engineering?**

“Building software in a structured, reliable, scalable, maintainable, and collaborative way.”

* The important thing is:

    * Not just coding
    * Not just building features
    * But building software that survives real-world usage



# **Core Goals of Software Engineering**

Students MUST understand these deeply because interviewers care about them.

## **1. Reliability**

Software should work correctly consistently.

> Example:

* Banking transaction systems
* Medical systems
* AI fraud detection systems

## **2. Maintainability**

Code should be easy to modify later.

> Important because:

* Most software lives for years
* Multiple engineers work on it
* Requirements change constantly

> Bad maintainability:

* Spaghetti code
* Tight coupling
* No documentation

> Good maintainability:

* Modular design
* Clear naming
* Layered architecture Tests

## **3. Scalability**

Software should handle growth.

> Growth can be:

* More users
* More requests
* More data
* More AI inference calls

## **4. Efficiency**

Efficient usage of:

* CPU
* Memory
* Network
* GPU
* Time

> AI relevance:

* Model inference optimization
* Vector database retrieval speed
* GPU utilization

## **5. Security**

Protect systems and user data.

> Examples:

* Authentication
* Authorization
* Encryption
* API security


## **6. Usability**

* Software should be understandable and easy to use.

* Even powerful software fails if users cannot use it properly.

## **Why Software Engineering Exists?**

Before software engineering:

* People wrote code individually
* Projects became chaotic
* Systems failed frequently
* Maintenance became impossible

Software Engineering emerged because software became:

* Large
* Complex
* Team-based
* Business-critical

# **❌ What problems occur without software engineering?**

Without software engineering, systems become messy, unstructured, and hard to maintain or scale. Bugs increase and teams can’t collaborate efficiently, leading to unstable products.

# **📈 Why is scalability important?**

Scalability ensures the system can handle growing users/data without slowing down or crashing. It lets the product grow smoothly without rewriting the whole system.

# **🛡️ What makes software reliable?**

Reliable software handles failures gracefully, has proper testing, monitoring, and can recover from errors quickly. It consistently works as expected under different conditions.

# **🤖 What software engineering problems appear in ML systems?**

ML systems face issues like data drift (Data drift means the input data changes over time compared to the data the model was trained on), model versioning, inconsistent training vs production data, and heavy compute/resource usage. They also need continuous monitoring after deployment.

# **🧠 Difference between building a model and building an AI product?**

Building a model means training an algorithm to perform a task on data. Building an AI product means integrating that model into a full system with UI, APIs, scaling, monitoring, and real users.

# **Interview angle:**

## **Q: “How do you ensure reliability in production systems?”**

### **Answer:**

1. “In production, I ensure reliability by designing systems that expect failure and recover gracefully.

2. On the engineering side, I use timeouts, retries with exponential backoff, idempotent APIs, and proper error handling to avoid cascading failures.

3. I rely heavily on observability—logs, metrics, tracing, and alerting—to quickly detect and diagnose issues in production.

4. For deployment safety, I use CI/CD with canary or blue-green releases and always have rollback mechanisms in place.

5. I also make systems degrade gracefully, so if a dependency like a model service or database slows down or fails, the system can fall back to cached or simpler responses instead of breaking completely.

Finally, I monitor post-deployment behavior closely, including latency, error rates, and data quality, especially in AI systems where model drift or pipeline issues can affect reliability.”

The goal is not just preventing failures, but recovering quickly and learning from them.”


## **Q: A chatbot serving 10 users vs 10 million users. How would you scale this system?”**

## **Answer:**

“If I scale a chatbot from 10 users to 10 million users, I focus on four main areas: compute, architecture, data, and reliability.

1. First, I move from a single app setup to a distributed system. I put the backend behind a load balancer so traffic is spread across multiple servers instead of one machine getting overloaded.

2. Second, I introduce horizontal scaling. Instead of upgrading one big server, I run multiple instances of the chatbot service in containers (like Kubernetes) so I can add/remove servers based on traffic.

3. Third, I optimize the AI/LLM layer because that becomes the bottleneck:

* I add caching for repeated queries
* I use batching for inference
* I may route simple queries to smaller/cheaper models and complex ones to bigger models
* I also use rate limiting to protect the system

4. Fourth, I introduce asynchronous processing where possible. Instead of making every request wait for heavy processing, I use queues (like Kafka/RabbitMQ) so requests are handled smoothly under load.

5. Fifth, I scale the data layer:

* use database sharding or read replicas
* optimize indexes
* possibly move to distributed databases or vector databases for embeddings

6. Sixth, I ensure reliability and observability:

* monitoring latency, errors, GPU usage
* autoscaling based on traffic
* circuit breakers so one failing service doesn’t crash everything

**Finally, I optimize cost because at 10 million users, efficiency matters:**

* caching responses
* compressing prompts
* using smaller models when possible
* avoiding unnecessary LLM calls

Overall, the goal is to shift from a single-system mindset to a distributed, fault-tolerant, and auto-scaling architecture that can handle unpredictable traffic.”

# **Q: Difference between authentication and authorization?**

## **Answer:**

> In simple terms, authentication is the process of verifying who the user is, usually through login credentials like username/password, OTP, or tokens.

> Authorization happens after authentication and determines what the user is allowed to do in the system, like access certain pages, APIs, or admin features.

So basically, authentication is identity verification, while authorization is permission control.”

* Authentication = who you are
* Authorization = what you can do

# **------------ Appendices ------------**

## **🔁 Retries with exponential backoff**

Agar koi request fail ho jaye (e.g. API down), system turant bar-bar hit nahi karta.
> Instead:

* 1st retry: thori der baad
* 2nd retry: zyada delay
* 3rd retry: aur zyada delay

👉 Matlab: “slowly try again so system overload na ho”

## **🔐 Idempotent APIs**

Same request agar 2–3 dafa bhi chali jaye, result same hi rahe, duplicate problem na ho.

> Example:

“Pay 100$” request accidentally 2 baar chali jaye

👉 idempotent system = sirf 1 payment ho

## **⚠️ Proper error handling (cascading failures)**

Agar 1 service fail ho jaye aur uski wajah se sab services gir jayein → usko cascading failure kehte hain.

👉 Fix: system ko aise design karna ke 1 part fail ho to baqi system na gire.

## **🚀 CI/CD + Canary / Blue-green release**

CI/CD: code automatically test + deploy hota hai.

> Canary release:

* pehle 1% users ko new version milta hai
* agar sab ok → phir sab ko roll out

> Blue-green deployment:

* old system (blue) chal raha hota hai
* new system (green) ready hota hai
* switch kar dete hain

👉 Matlab: “pehle chhota test, phir full launch”

## **🔄 Rollback mechanism**

Agar new version break ho jaye:

👉 1 click / auto → purana stable version wapas

## **🧠 System degrade gracefully (LLM fallback)**

Agar AI/LLM service fail ho jaye:

> Instead of app crashing:

* cached answers use kar leta hai (pehle se saved replies)
* ya simpler model use karta hai
* ya basic rule-based response deta hai

👉 Matlab: “perfect answer na sahi, but system still chalti rahe”

## **📉 Model drift**

Time ke sath AI model “purana” ho jata hai because:

* data change ho gaya
* user behavior change ho gaya

> Example:

* model ne 2024 data pe train kiya tha
* 2026 me new trends aa gaye

👉 Result: model ki accuracy dheere dheere girti hai

## **🔹 Horizontally scalable**

Matlab system ko scale karne ke liye ek heavy server ko bada nahi karte, balkay aur servers add kar dete hain.
Traffic multiple machines pe divide ho jata hai.
👉 Simple: “1 machine ki jagah 10 machines laga do”

## **🔹 Kafka (simple idea)**

Kafka ek message system / pipeline hai jo data ko temporarily store aur pass karta hai between services.
Ek service message bhejti hai, doosri baad me le leti hai.
👉 Simple: “data ka middleman jo messages safely carry karta hai”

## **🔹 Queue system (kya hota hai)**

Queue matlab line mein lagna — requests ek order mein wait karti hain processing ke liye.
System ek ek karke handle karta hai instead of overload.
👉 Simple: “requests ka waiting line system”


# **🧩 Database sharding**

Database ko chhote parts (shards) mein divide karna.
Har part different server par hota hai aur data ka ek hissa handle karta hai.
👉 Simple: “1 big database ko tod kar multiple databases bana dena”

# **📖 Read replicas**

Main database ki copies bana dete hain sirf read (data dekhne) ke liye.
Write main database me hota hai, reads replicas se hoti hain.
👉 Simple: “read load distribute karne ke liye copies bana lena”

# **⚡ Index optimization**

Index database ka “shortcut” hota hai jisse search fast hoti hai.
Optimize karna matlab sahi columns pe index banana aur unnecessary indexes remove karna.
👉 Simple: “database search ko fast karne ka smart shortcut set karna”

# **🔌 Circuit breaker**

Agar ek service bar-bar fail ho rahi ho, system temporarily usko call karna band kar deta hai.
Taake poora system slow ya crash na ho jaye.
👉 Simple: “faulty service ko temporarily disconnect kar dena”

# **🛡️ Fault-tolerant system**

System aisa hota hai jo kuch parts fail ho jayein phir bhi chalta rahe.
Dusre backups ya fallback use karke service continue karta hai.
👉 Simple: “failure ke bawajood system ka chalte rehna”

# **How we can make an API idempotent?**

An API is idempotent when calling it multiple times with the same request gives the same result without unwanted side effects (like duplicates).
Here’s how we make an API idempotent:

🔑 1. Use unique request IDs (Idempotency key)
Client sends a unique key with each request.
Server checks: if already processed → return same response, don’t re-run.

🧾 2. Check before creating/updating
Before inserting data, check if it already exists (based on user/order ID).
Avoid duplicate creation.

🔁 3. Use UPSERT operations
Instead of “insert only”, use insert-or-update logic in DB.
So repeated calls just overwrite same record safely.

🔒 4. Store request state
Save processed request IDs in database/cache.
If same request comes again → return previous result.

🔥 Interview line:
“We make an API idempotent by ensuring repeated requests with the same input don’t create duplicate side effects, using techniques like idempotency keys, deduplication checks, and UPSERT operations.”

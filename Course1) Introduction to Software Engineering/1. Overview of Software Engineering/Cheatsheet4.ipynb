# **Senior AI Enginner MCQs**

## **1. What best defines Software Engineering?**

A. Writing code only

B. Designing and maintaining software using engineering principles

C. Testing software after development

D. Deploying applications

Answer: B

## **2. Which is NOT part of the SDLC?**

A. Requirements Analysis

B. Deployment

C. Marketing

D. Maintenance


Answer: D

## **3. The main purpose of SDLC is to:**

A. Increase coding speed


B. Deliver software in a structured and efficient way

C. Replace developers with automation

D. Avoid testing

Answer: B

## **4. Which SDLC phase defines “what to build”?**

A. Design

B. Testing

C. Requirements

D. Deployment

Answer: C

## **5. Which SDLC model is most flexible and iterative?**

A. Waterfall

B. V-Model

C. Agile

D. Big Bang

Answer: C

## **6. What is the correct order of SDLC phases?**

A. Design → Requirements → Testing → Deployment

B. Requirements → Design → Development → Testing → Deployment

C. Testing → Requirements → Design → Deployment

D. Deployment → Development → Requirements → Testing

Answer: B

## **7. Insiders' viewpoint of Software Engineering mainly focuses on:**

A. Tools and languages only

B. Human collaboration, trade-offs, and constraints

C. Only coding practices

D. Only testing strategies

Answer: B

## **8. Agile methodology is primarily based on:**


A. Heavy documentation

B. Fixed long-term planning

C. Iterative development and feedback

D. No planning at all

Answer: C


## **9. In Agile, the most important output is:**

A. Documentation

B. Working software

C. UML diagrams

D. Test plans

Answer: B


## **10. A sprint in Agile is:**

A. Final release phase


B. Short development cycle with deliverable output

C. Bug fixing phase

D. Requirement gathering phase

Answer: B
